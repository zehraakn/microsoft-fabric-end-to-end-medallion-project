```python
file_name = '2016Feb.csv'
processed_date = '2025-12-30'
```

```python
adfs_path='abfss://Fabric_Dev@onelake.dfs.fabric.microsoft.com/Fabric_LH_Sales.Lakehouse/Files/raw'
```

```python
full_path= f"{adfs_path}/{file_name}"
print(full_path)
```

```python
df = spark.read.csv(
    path=full_path,
    header=True,
    inferSchema=True,
    sep=";"
)

display(df)
```

```python
from pyspark.sql.functions import lit
if df.count()>1:
    print("This file has data")
    df_new = df.withColumn("Processing_date",lit(processed_date))
    #display(df_new)
    df_new.write.format('csv').option("header","True").mode("append").partitionBy("Processing_date").save("abfss://Fabric_Dev@onelake.dfs.fabric.microsoft.com/Fabric_LH_Sales.Lakehouse/Files/Landing")
    print("Data Loading to Landing zonw successfuly")
else:
    print("File contain no Data")
```
